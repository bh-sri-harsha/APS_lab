```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)

#population parameters for normal distribution
population = np.random.normal(loc=0, scale=1, size=1000)
pop_mean = np.mean(population)
pop_std = np.std(population)

#sample parameters
sample_indices = np.random.choice(len(population), size=100, replace=False)
sample = population[sample_indices]

#mean, standard deviation, and standard error of the sample
sample_mean = np.mean(sample)
sample_std = np.std(sample,ddof=1)
std_error = sample_std / np.sqrt(len(sample))
```

```python
print(f"Population Mean: {pop_mean:.4f}, Population Std Dev: {pop_std:.4f}")
```
