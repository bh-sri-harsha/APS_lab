# Problem Set 1: Simulating Customer Clicks Using a Bernoulli Random Variable

## Problem statement
An organization sends promotional emails. A customer clicks the link with probability `p`.

Define a Bernoulli random variable:
- `X = 1` if the customer clicks.
- `X = 0` if the customer does not click.

**Source note:** The uploaded task document appears to omit the numerical value of `p`. Enter the value supplied by your instructor when prompted below; do not treat any default value as given by the source.

## Tasks
1. Identify the random variable and its distribution.
2. Write the theoretical PMF.
3. Simulate outcomes for 20 customers.
4. Count clicks and non-clicks.
5. Calculate the empirical PMF.
6. Compare empirical and theoretical PMFs.
7. Plot both PMFs.
8. Calculate theoretical and empirical expectation.
9. Calculate theoretical and empirical variance.
10. Prepare a comparison table.
11. Repeat for sample sizes 10, 100, 1,000, and 10,000.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Reproducibility: the same seed gives the same simulated results.
rng = np.random.default_rng(42)

p = float(input("Enter the customer click probability p (between 0 and 1): "))
if not 0 <= p <= 1:
    raise ValueError("p must be between 0 and 1.")

sample_size = 20
```

## Step 1 — Identify the random variable

```python
print("X = 1 means the customer clicks.")
print("X = 0 means the customer does not click.")
print("Possible values: {0, 1}")
print("X is discrete.")
print("Suitable distribution: Bernoulli(p).")
```

## Step 2 — Theoretical PMF

For a Bernoulli random variable:

- \(P(X=0)=1-p\)
- \(P(X=1)=p\)

```python
outcomes = np.array([0, 1])
theoretical_pmf = np.array([1 - p, p])

pd.DataFrame({
    "Outcome x": outcomes,
    "Theoretical P(X=x)": theoretical_pmf
})
```

## Step 3 — Generate outcomes for 20 customers

```python
samples = rng.binomial(n=1, p=p, size=sample_size)
print("Generated observations:", len(samples))
print("Samples:", samples)
print("0 = no click; 1 = click.")
print("The sample proportions need not exactly equal the theoretical probabilities.")
```

## Step 4 — Count clicks and non-clicks

```python
number_of_clicks = np.sum(samples == 1)
number_of_non_clicks = np.sum(samples == 0)

print("Number of clicks:", number_of_clicks)
print("Number of non-clicks:", number_of_non_clicks)
print("Counts add up to 20:", number_of_clicks + number_of_non_clicks == sample_size)
```

## Step 5 — Empirical PMF

```python
empirical_pmf = np.array([
    np.mean(samples == 0),
    np.mean(samples == 1)
])

pd.DataFrame({
    "Outcome x": outcomes,
    "Theoretical PMF": theoretical_pmf,
    "Empirical PMF": empirical_pmf
})
```

## Step 6 — Compare empirical and theoretical PMFs

```python
pmf_comparison = pd.DataFrame({
    "Outcome x": outcomes,
    "Theoretical PMF": theoretical_pmf,
    "Empirical PMF": empirical_pmf,
    "Absolute difference": np.abs(theoretical_pmf - empirical_pmf)
})
display(pmf_comparison)

print("Is empirical click probability exactly p?", np.isclose(empirical_pmf[1], p))
print("The larger theoretical-probability outcome is:",
      outcomes[np.argmax(theoretical_pmf)])
print("Differences occur because a finite random sample varies from the theoretical distribution.")
```

## Step 7 — Plot empirical and theoretical PMFs

```python
x = np.arange(2)
width = 0.35
plt.bar(x - width/2, theoretical_pmf, width, label="Theoretical PMF")
plt.bar(x + width/2, empirical_pmf, width, label="Empirical PMF")
plt.xticks(x, ["0 (no click)", "1 (click)"])
plt.ylabel("Probability")
plt.title("Theoretical vs Empirical Bernoulli PMF")
plt.legend()
plt.tight_layout()
plt.show()
```

## Steps 8–11 — Expectation and variance

```python
# Theoretical Bernoulli moments
theoretical_mean = p
theoretical_variance = p * (1 - p)

# Empirical moments from the generated observations
empirical_mean = np.mean(samples)
empirical_variance = np.var(samples, ddof=0)  # population-style empirical variance

result_table = pd.DataFrame({
    "Measure": ["P(X=0)", "P(X=1)", "Expectation", "Variance"],
    "Theoretical value": [1-p, p, theoretical_mean, theoretical_variance],
    "Empirical value": [empirical_pmf[0], empirical_pmf[1], empirical_mean, empirical_variance]
})
result_table["Absolute difference"] = np.abs(
    result_table["Theoretical value"] - result_table["Empirical value"]
)
result_table
```

## Step 13 — Effect of sample size

```python
rows = []
for n in [10, 100, 1_000, 10_000]:
    observations = rng.binomial(n=1, p=p, size=n)
    rows.append({
        "Sample size": n,
        "Empirical P(X=0)": np.mean(observations == 0),
        "Empirical P(X=1)": np.mean(observations == 1),
        "Empirical mean": np.mean(observations),
        "Empirical variance": np.var(observations, ddof=0)
    })

sample_size_table = pd.DataFrame(rows)
sample_size_table
```

### Interpretation
As the sample size grows, empirical probabilities, mean, and variance will generally tend to be closer to their theoretical values. Random variation means the results are not guaranteed to improve monotonically at every sample size.
