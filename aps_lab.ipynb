1.	Problem1: A fair coin is tossed 100 times.
Calculate:

•	Number of heads 

•	Number of tails

•	Experimental probability of Head 

•	Experimental probability of Tails


```python
toss = int(input("Enter no. of tosses boss"))
he = rand.randint(1,toss)
ta = toss - he
print(f"No. of tosses is {toss}")
print(f"No. of head {he}")
print(f"No. of tails {ta}")
head_prob = he/toss
tail_prob = ta/toss
print(f"Experimental prob. of heads for {toss} tosses of coin is {head_prob}")
print(f"Experimental prob. of tails for {toss} tosses of coin is {tail_prob}")
```

4.	Addition rule for disjoint events

A die is rolled 10,000 times.

Define: A=getting 1, B= getting 6, check the events are joint or disjoint 

Verify: P(A∪B)=P(A)+P(B)


```python
rolls = 10000
se = {1,2,3,4,5,6}
prob_a = 1/len(se)
prob_b = 1/len(se)
prob_a_intersec_b = 0
prob_a_union_b = prob_a + prob_b - prob_a_intersec_b
booli = True if prob_a_union_b == prob_a + prob_b else False
print(f"P(A∩B) is {prob_a_intersec_b},therefore set is disjoint")
print(f"P(A∪B)=P(A)+P(B) is {booli}")
```

5.	Addition rule for overlapping events

For one die roll, define: A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6}

Verify: P(A∪B)=P(A)+P(B)−P(A∩B)


```python
A = {2,4,6}
B = {4,5,6}
A_int_B = {4,6}
se ={1,2,3,4,5,6}
prob_a = len(A)/len(se)
prob_b = len(B)/len(se)
prob_A_int_B = len(A_int_B)/len(se)
prob_A_uni_B = prob_a + prob_b - prob_A_int_B
print(f"P(A∩B) is {prob_A_int_B}")
print(f"P(A∪B) is {prob_A_uni_B}")
booli = True if prob_A_uni_B == prob_a + prob_b - prob_A_int_B else False
print(f"P(A∪B)=P(A)+P(B)−P(A∩B) is {booli}")
```

6.	Multiplication rule for independent events

A coin is tossed and a die is rolled 10,000 times.

Define:

•	obtaining Head 
•	obtaining an even number

Verify: P(A∩B)=P(A)P(B)


```python
toss = int(input("Enter no. of tosses boss"))
he = rand.randint(1,toss)
prob_a = he/toss
se = {1,2,3,4,5,6}
B = {2,4,6}
prob_b = len(B)/len(se)
p_a_int_b = prob_a * prob_b
print(f"P(A∩B)=P(A)P(B) is verified as {p_a_int_b} is equal to {prob_a * prob_b}")
```
